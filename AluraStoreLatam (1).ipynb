# 4. Productos más y menos vendidos

```python
def analizar_tiendas(lista_dfs, nombres):
    resumen = []
    for df, nombre in zip(lista_dfs, nombres):
        # Contar ventas y obtener el más/menos vendido
        conteo = df['Producto'].value_counts()
        p_top, p_low = conteo.idxmax(), conteo.idxmin()

        # Guardar datos buscando la categoría directamente
        resumen.append({
            "Tienda": nombre,
            "Producto Estrella": p_top,
            "Categoría": df[df['Producto'] == p_top]['Categoría del Producto'].iloc[0],
            "Ventas": conteo.max(),
            "Producto Menos Vendido": p_low,
            "Ventas Min": conteo.min()
        })

    return pd.DataFrame(resumen)

mis_tiendas = [tienda1, tienda2, tienda3, tienda4]
mis_nombres = ["Tienda 1", "Tienda 2", "Tienda 3", "Tienda 4"]

df_final = analizar_tiendas(mis_tiendas, mis_nombres)
display(df_final)
```

## 1. Introducción
Este informe tiene como objetivo identificar la mejor oportunidad de negocio para el Sr. Juan entre cuatro tiendas analizadas. Se evaluaron métricas de ingresos, costos de envío y satisfacción del cliente, calificaciones para garantizar una decisión basada en datos y no en suposiciones.

```python
# @title

tiendas_list = [tienda1, tienda2, tienda3, tienda4]
nombres = ["Tienda 1", "Tienda 2", "Tienda 3", "Tienda 4"]
datos_informe = []

for i, df in enumerate(tiendas_list):

    ingreso = df['Precio'].sum()
    envio = df['Costo de envío'].mean()
    rating = df['Calificación'].mean()

    conteo = df['Producto'].value_counts()
    p_top = conteo.idxmax()
    p_low = conteo.idxmin()


    cat_estrella = df[df['Producto'] == p_top]['Categoría del Producto'].iloc[0]

    datos_informe.append({
        "Tienda": nombres[i],
        "Ingreso Total": ingreso,
        "Envío Prom.": envio,
        "Rating": rating,
        "Producto Estrella": p_top,
        "Categoría Estrella": cat_estrella,
        "Ventas (Max)": conteo.max(),
        "Producto menos vendido": p_low
    })

df_final = pd.DataFrame(datos_informe).sort_values(by="Ingreso Total", ascending=False)


print(" TABLA DESCRIPTIVA DE LOS DATOS DE CADA TIENDA")
display(df_final.style.format({
    "Ingreso Total": "${:,.2f}",
    "Envío Prom.": "${:,.2f}",
    "Rating": "{:.2f}"
}))
```

***Conclusiones***

Tras verificar las métricas de las cuatro sucursales, la recomendación técnica es proceder con la **liquidación o venta de la Tienda 4**.

Esta decisión se basa en el análisis de los siguientes indicadores de desempeño (**KPIs**):

* **Bajo Volumen de Ventas:** La Tienda 4 presenta el **Ingreso Bruto** más bajo de todo el grupo. A pesar de operar en el mismo mercado, su capacidad de captación de capital es inferior a las demás.
* **Mix de Productos Ineficiente:** La tienda depende de categorías con bajos márgenes de ganancia. No ha logrado posicionarse en los sectores de **Alta Rotación** (como Electrónicos), que son los que realmente sostienen la rentabilidad del negocio.
* **Ventaja Logística Desaprovechada:** Aunque tiene el **Costo de Envío** más económico, este ahorro no es suficiente para compensar la falta de ventas. En términos técnicos: la eficiencia en costos no compensa la debilidad en los ingresos.
* **Reputación Estancada:** Su **Rating (Calificación)** no destaca sobre la competencia, lo que significa que no aporta un valor de marca diferencial que justifique mantenerla operativa.

**Veredicto:** El Sr. Juan debe desinvertir en la **Tienda 4** para evitar la fuga de capital y concentrar sus recursos en las tiendas líderes, asegurando así la sostenibilidad de su patrimonio.
